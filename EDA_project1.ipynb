```python
#outliers
#q1,q2,q3,IQR
import pandas as pd
import numpy as np
Q1=df["Age"].quantile(0.25)
Q3=df["Age"].quantile(0.75)
IQR = Q3-Q1
print("Q1:" ,Q1)
print("Q3:" ,Q3)
print("IQR:" ,IQR)

lower_bound = (Q1-1.5*IQR)
upper_bound = (Q3+1.5*IQR)
print(lower_bound)
print(upper_bound)

#identify IQR outliers
IQR_outliers=df[(df["Age"]<lower_bound)|(df["Age"]>upper_bound)]
print("IQR OUTLIERS:",len(IQR_outliers))
```

```python
#methods
#1.zscore method:- to bring back an outlier as normal point. the outlier should justify
from scipy.stats import zscore
df["Age_zscore"]= zscore(df["Age"])
z_outliers = df[df["Age_zscore"].abs() > 3]
print("z-score outliers:", len(z_outliers))
```

```python
#z-score trimming
df_trimmed=df[df["Age_zscore"].abs() <=3 ]
print(df_trimmed)
```

```python
#winsorization
from scipy.stats.mstats import winsorize
df['Age_winsorized']=winsorize(df["Age"],limits=[0,0.3])
print(df.head())
```
